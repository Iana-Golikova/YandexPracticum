## Анализ

###  Количестов книг, вышедших после 1 января 2000 года

Посчитайтаем, сколько книг вышло после 1 января 2000 года:

```python
query = '''SELECT  COUNT(*)
FROM books 
   WHERE publication_date>'2000-01-01' '''
pd.io.sql.read_sql(sql=text(query), con = con)
```

После 1 января 2000 года вышло 819 книг.

### Средняя оценка и количество обзоров

Для каждой книги посчитаем количество обзоров и среднюю оценку.

```python

query = '''
SELECT rv.book_id, rv.title,  rv.cnt, rt.avg_rate FROM
    (SELECT b.book_id, b.title, COUNT(r.review_id) AS cnt 
        FROM books b
        LEFT JOIN reviews r ON b.book_id = r.book_id
        GROUP BY b.book_id,b.title
        
    ) AS rv
    JOIN 
    (SELECT b.book_id, AVG(r.rating) AS avg_rate
        FROM books b
        LEFT JOIN ratings r ON b.book_id = r.book_id
        GROUP BY b.book_id
        ORDER BY  avg_rate
    ) AS rt ON rv.book_id = rt.book_id '''
pd.io.sql.read_sql(sql=text(query), con = con)
```

Средняя оценка книг находится в диапазоне от 3 до 4.5. Количество обзоров от 0 до 7.

### Самое крупное издательство книнг

Определим издательство, которое выпустило наибольшее число книг толще 50 страниц — так мы исключим из анализа брошюры.

```python
query = '''
SELECT p.publisher, COUNT(*) 
    FROM publishers  p
    JOIN books b ON p.publisher_id = b.publisher_id
    WHERE b.num_pages>50
    GROUP BY p.publisher_id, p.publisher
    ORDER BY COUNT(*) DESC
    LIMIT 1'''
pd.io.sql.read_sql(sql=text(query), con = con)
```

Самое крупное книжное издательство — "Penguin Books". Оно выпустило 42 книги, объемом более 50 страниц

### Самый высокорейтинговый автор

Определим автора с самой высокой средней оценкой книг. 

```python
query = '''
SELECT a.author, AVG(bs.rate)
    FROM authors a
    JOIN 
    (SELECT b.book_id, b.author_id, AVG(r.rating) as rate
        FROM books b
        JOIN ratings r ON b.book_id = r.book_id
        GROUP BY b.book_id, b.author_id
        HAVING COUNT(*)>=50
    ) AS bs ON a.author_id = bs.author_id
    GROUP BY a.author_id, a.author
    ORDER BY AVG(bs.rate) DESC
    LIMIT 1'''
pd.io.sql.read_sql(sql=text(query), con = con)
```

Автор с самыми высокой средней оценкой книг — GrandPré. Средняя оценка 4.3.

### Среднее количество обзоров пользователей

Посчитаем среднее количество обзоров от пользователей, которые поставили больше 48 оценок.

```python
query = '''
SELECT AVG(rv.cnt) 
    FROM
    (SELECT COUNT(*) as cnt
        FROM reviews rv
        JOIN
        (SELECT username
            FROM ratings
            GROUP BY username
            HAVING COUNT(*) > 48
        ) u on rv.username = u.username
         GROUP BY u.username
    ) AS rv'''
pd.io.sql.read_sql(sql=text(query), con = con)
```

Среднее количество обзоров у активных пользовтаелей — 24.
