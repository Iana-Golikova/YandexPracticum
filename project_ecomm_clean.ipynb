## Вспомогательные функции, используемые в проекте.

Зададим вспомогательные общие функции, которые неоднокртано будтут использоваться в проекте.

```python
# функция, создающая сводную таблицу

def create_pivot_table (data, index_list, values_list, aggfunc_list, reset=True):
    new_table=data.pivot_table(index = index_list, values = values_list, aggfunc = aggfunc_list)
    new_table.columns=list(map(lambda x: x[0]+'_'+x[1], new_table.columns)) # переименовываем столбцы
    if reset == True:
        new_table=new_table.reset_index() # ресет индексов
    return new_table
```

```python
# функция, строящая график распределения
def create_kdeplot(data, column, size, xlabel, title, log_Flag = False):
    col= data[column]
    if log_Flag == True:
        col = np.log(col)
  
    plt.subplots(figsize = size)
    plt.xlabel(xlabel)
    plt.ylabel('плотность распределения')
    plt.title('Распределение '+title)
    sns.kdeplot(col, fill = True, common_grid = True)
    plt.grid()
```

```python
# функция, строящая гистограмму

def create_hist (data, col, bins, size, header, xlab, ylab) :
    ax = data.plot(y = col, bins=bins, figsize = size, kind = 'hist',
                   title = header, legend=False, grid = True);
    ax.set_xlabel(xlab)
    ax.set_ylabel(ylab)
```

```python
# функция, строящая линейный график

def create_line_plot (data, xcol, ycol, size, header, xlab, ylab,xtiks= True) :
    ax = data.plot(x = xcol, y = ycol, figsize = size, kind = 'line',marker='o',
                   title = header, legend=False, grid = True);
    if xtiks == True:
        plt.xticks(ticks=range(len(data[xcol])), labels=data[xcol])
    ax.set_xlabel(xlab)
    ax.set_ylabel(ylab)
```

```python
# Построение линейного инетрактивного графика 
def create_line_px (data, xcol, ycol, width,height, header, xlab, ylab, color=None, color_map = None, xtiks= False) :
    fig = px.line(data, x=xcol, y= ycol, # строим сам график
                  title=header ,template='plotly_white', markers= True,
                  color=color, color_discrete_map = color_map)
    
    fig.update_layout(xaxis_title=xlab ,
                      yaxis_title=ylab,
                      width = width,   # Ширина графика в пикселях
                      height = height,
                      yaxis=dict(range=[0, data[ycol].max()*1.05])
                     )
    if xtiks == True: # настраиваем сетку
        fig.update_xaxes(tickmode='array', tickvals=data[xcol])
    fig.show()

```

```python
# функция постороения барплота
def create_bar (data,xcol, ycol,title,xlable,ylable,color = None,color_map = None):
    fig = px.bar(data.sort_values(by=ycol, ascending=False), #сам столбчатый график
             x= xcol,
             y= ycol,
             color = color, 
             color_discrete_map = color_map, # задаем цветовую карту, чтобы на однотипных графиках категории были одного цвета
             text= ycol,
             title=title,
             template='plotly_white'
            )
    fig.update_layout(xaxis_title=xlable, # подписываем оси
                  yaxis_title=ylable
                  )
    fig.show()
    
```

```python
# функция создания круговой диаграммы
def create_pie(data, value, title):
    pie_1=px.pie(data,  # создаем саму диаграмму
                 values = value, 
                 names ='category',
                 width = 800, height=660,
                 color = 'category',
                 color_discrete_map=color_map
                 )
    pie_1.update_traces(textposition='inside', textinfo='percent+label+value') #подписи внутри диаграммы
    pie_1.update_layout(title=title, # название+легенда
                        annotations=[dict(x=1.10, 
                                    y=1.05,
                                    text='Категория',
                                    showarrow=False)])
    pie_1.show()
```

Прежде чем вносить дальнешие изменения, создадим копию датасета и дальше будем работать с ней. 

```python
ecom_data = original_ecom.copy()
```

### Провекра на наличие дубликатов. 

Проверим на явные дубликаты:

```python
print(ecom_data.duplicated().sum())
```

Явных дублиактов нет, но могут быть неявные.

### Добавление столбцов.

Добавим необходимые столбцы

```python
# выручка заказа
ecom_data['revenue'] = ecom_data['quantity']*ecom_data['price'] 

# день заказа
ecom_data['date_order']=ecom_data['date'].dt.date

# год
ecom_data['year']=ecom_data['date'].dt.year

#месяц с годом
ecom_data['month']=ecom_data['date'].dt.to_period('M')
ecom_data['month'] = ecom_data['month'].dt.strftime('%Y-%m')

# день недели 0-пн
ecom_data['day_of_week']=ecom_data['date'].dt.dayofweek

# час
ecom_data['hour']=ecom_data['date'].dt.hour


display(ecom_data.head())
```

### Выявление неявных дубликатов и аномальных значений.

Приведем названия продуктов в нижний регистр и заменим `ё` на `е`.

```python
# приведем значения в столбце к нижнему регистру

ecom_data['product'] = ecom_data['product'].str.lower()

# заменим букву ё на е

ecom_data['product'] = ecom_data['product'].str.replace('ё','е')

display(ecom_data.head(50))
```

Каждому заказу должен соотвествовать уникальный идентификатор клиента. Проверим, есть ли аномальные заказы, которым соответсвует несколько идентификаторов клиента.

```python
# группируем по номеру заказа и считаем уникальных пользователей

ecom_data_group_order_id = ecom_data.groupby('order_id').agg({"customer_id": ['nunique']})\
                                                        .sort_values(by=[('customer_id', 'nunique')], ascending=False)

ecom_data_group_order_id = ecom_data_group_order_id[ecom_data_group_order_id[('customer_id', 'nunique')]>1]

display(ecom_data_group_order_id)
```

```python
# создаем список заказов, которым соотвествует более одного клиента

non_unique_orders_list = ecom_data_group_order_id.index.tolist()

print("Всего заказов с более чем одним клиентом:",len(non_unique_orders_list))
```

```python
# полностью удаляем заказы более чем с одним клиентом

ecom_data.drop(ecom_data[ecom_data['order_id'].isin(non_unique_orders_list)].index, inplace=True)
```

Таже каждому заказу должны соответсвовать единственные дата и время.

## Категоризация товаров.

### Создание категорий и словарей.

Категоризуем товары. Для этого выведем все уникальные названия продуктов.

```python
print(ecom_data['product'].sort_values().unique().tolist())
```

Выделим следующие категории:
* растения - `plant`;
* товары для дома - `houseware`;
* посуда (товары для кухни) - `dishes`;
* декор - `decoration`;
* мебель - `construction`.

```python
# функция категоризцации
def categorize_text(df: pd.DataFrame, text_column, category_dict, default_category, output_column):
    def get_category(text):
        if isinstance(text, str):
            for category, prefixes in category_dict.items():
                if any(text.startswith(prefix) for prefix in prefixes):
                    return category
        return default_category
    
    df[output_column] = df[text_column].apply(get_category)
```

## Взаимосвязь между заказами одного пользователя.

Проанализируем взаимосвязи между заказами одного пользователя. Для этого создадим таблицу со средним чеком пользователя, маркером нескольких покупок, списком категорий из первого заказа, списком категорий из второго заказа и периода между заказами

```python
# составляем список категорий в заказе
def concat_distinct(series): 
    return ','.join(series.sort_values().unique())
    

ecom_data_pivot_by_order=create_pivot_table(ecom_data,['order_id'],  
                                            ['customer_id', 'product','date_order','quantity','revenue','category'],
                                            {'customer_id': [ 'first'],
                                             'product': [ 'count'],
                                             'date_order': ['first'], 
                                             'quantity':['sum'],
                                             'revenue':['sum'],
                                             'category':[concat_distinct]})
```

```python
# ранжируем покупки
ecom_data_sorted = ecom_data_pivot_by_order.sort_values(by=['customer_id_first', 'date_order_first'])
ecom_data_sorted['order_rank'] = ecom_data_sorted.groupby('customer_id_first').cumcount() + 1
```

```python


# Объединяем даты первого и второго заказа
date_diff = ecom_data_sorted.merge(ecom_data_sorted, on='customer_id_first', how='inner', suffixes = ('_pre','_next'))

date_diff = date_diff[date_diff['order_rank_next']==date_diff['order_rank_pre']+1]

date_diff['date_order_first_next'] = pd.to_datetime(date_diff['date_order_first_next'])
date_diff['date_order_first_pre'] = pd.to_datetime(date_diff['date_order_first_pre'])


date_diff['time_diff_days'] = (date_diff['date_order_first_next'] - date_diff['date_order_first_pre']).dt.days

display(date_diff.head())
display(date_diff[['time_diff_days',
                   'category_concat_distinct_pre',
                   'revenue_sum_pre',
                   'category_concat_distinct_next',
                   'revenue_sum_next']])
create_hist(date_diff[date_diff['category_concat_distinct_pre']==date_diff['category_concat_distinct_next']],
            'time_diff_days', 50, (15, 10),
            'Гистограмма количества дней между последовательными заказами одного пользователя', 
            'количество дней', 
            'количестово пар заказов')
```

Есть явное разделение между покупателями, возвращающимся в ближайшее время (примерно первый месяц) и остальными. Динамкика возвращений за первые два месяца убывает экспоненциально. Динамика возвращения покупателей после трех месяцев хаотична.
Так как период рассмотрения всего год с небольшим, о динамике долгосрочного возвращения судить сложно.

```python
category_list_pre=date_diff['category_concat_distinct_pre'].unique().tolist()
category_list_next=date_diff['category_concat_distinct_next'].unique().tolist()
display(category_list_pre,category_list_next)
```

Напишем функцию, создающую таблицу для сегментации. 

```python
def aggregate_orders(df, dates, delta_past, delta_future):
    result = []
    
    all_users = df['customer_id_first'].unique()
     
    for now in dates:
        date_from = now - delta_past # начало окна "прошлого"
        date_to = now + delta_future # конец окна "будущего"
        
        
        # фильтруем по окнам
        filtered_df_past = df[(df['date_order_first'] > date_from) & (df['date_order_first'] <= now)]
        filtered_df_future = df[(df['date_order_first'] > now) & (df['date_order_first'] <= date_to)]
        
        # группируем по пользователю
        grouped_past = filtered_df_past.groupby('customer_id_first').agg(
            order_count = ('revenue_sum', 'count'),
            total_value = ('revenue_sum', 'sum'),
            last_order = ('date_order_first', 'max')
        ).reset_index()        
        grouped_future = filtered_df_future.groupby('customer_id_first').agg(
            order_count=('revenue_sum', 'count'),
            total_value=('revenue_sum', 'sum')
        ).reset_index()
        
        # переименовываем колонки
        grouped_past.columns = ['customer_id', 'past_cnt', 'past_sum', 'last_order']        
        grouped_future.columns = ['customer_id', 'fut_cnt', 'fut_sum']
        
        # считаем время с последней покупки 
        grouped_past['last_order_days_ago'] = grouped_past['last_order'].apply(lambda x: (now - x).days)
        
        # сливаем
        grouped = grouped_past.merge(grouped_future, how = 'left')

        # заполняем нули
        grouped = grouped.fillna(0.0)
        #grouped = grouped.set_index('customer_id').reindex(all_users, fill_value=0).reset_index().fillna(0.0)
        
        # добавляем дату
        grouped.insert(0, 'date', now)
        # сохраняем в список
        result.append(grouped)
    
    # схлопываем список в одну таблицу
    return pd.concat(result, ignore_index=True)
```

```python
# формируем список дат 'настоящее'

delta_past = dt.timedelta(days = 7*7) # 7 недель 
delta_future = dt.timedelta(days = 7*7) # 7 недель
step = dt.timedelta(days=7)

min_date = ecom_data['date_order'].min() + delta_past
max_date = ecom_data['date_order'].max() - delta_future
dates = []
date = min_date
while date <= max_date:
    dates.append(date)
    date = date + step
```

```python
# Явный список сегментов, будем его потом использовать для попарного сравнения 
segments = ['Много, давно', 'Много, недавно', 'Мало, давно', 'Мало, недавно', 'Повторные']

def get_stats(table):
    threshold_sum = table[(table['past_cnt']>0)]['past_sum'].median()
    threshold_days_ago = delta_past.days / 2 # можно тоже медиану, но середина интервала даёт достаточно равномерное разбиение
    table['segment'] = 'Нет покупок'
    table.loc[(table['past_cnt']==1)
              &(table['last_order_days_ago']>=threshold_days_ago)
              &(table['past_sum']>=threshold_sum), 'segment'] = 'Много, давно'
    
    table.loc[(table['past_cnt']==1)
              &(table['last_order_days_ago']>=threshold_days_ago)
              &(table['past_sum']<threshold_sum), 'segment'] = 'Мало, давно'

    table.loc[(table['past_cnt']==1)
              &(table['last_order_days_ago']<threshold_days_ago)
              &(table['past_sum']>=threshold_sum), 'segment'] = 'Много, недавно'

    table.loc[(table['past_cnt']==1)
              &(table['last_order_days_ago']<threshold_days_ago)
              &(table['past_sum']<threshold_sum), 'segment'] = 'Мало, недавно'
    
    table.loc[(table['past_cnt']>1),'segment'] = 'Повторные'

    stats = (create_pivot_table(table, 'segment', ['fut_cnt','fut_sum'],
                           {'fut_cnt':['count','mean','var'],'fut_sum':['mean','var']}, reset=False))
        
    stats.rename(columns = {'fut_cnt_count':' count'}, inplace = True)
    
    display(stats)
    
    return table[['segment','fut_cnt','fut_sum','date']]

common_stats = get_stats(table)
display(common_stats)
```

#### Проверка гипотез

Проверим гипотезы попрано по сегментам.

Первая гипотеза о различии выручки по сегментам:
* **Нулевая гипотеза $H_0$**: выручка на пользователя в будущем одинаковая для разных сегментов;
* **Альтернативная гипотеза $H_1$**: выручка на пользователя в будущем различна для разных сегментов;

Второая о различии количества заказов:
* **Нулевая гипотеза $H_0$**: количество заказов в будущем одинаковое для разных сегментов;
* **Альтернативная гипотеза $H_1$**: количество заказов в будущем различно для разных сегментов;

Проверка гипотезы о различии выручки по сегментах:


```python
# Считаем пороговую значимость с поправкой Бонферрони
h_cnt = len(segments)*(len(segments)-1)/2 # Количество пар сегментов
base_alpha = 0.05 # Стандартный уровень значимости
alpha = base_alpha / h_cnt # Уровень значимости с поправкой
```

```python
def compare_segments(data, column, column_description, accepted_hypotheses):
    values_by_seg = []
    for seg in segments:
        values_by_seg.append(data[data['segment']==seg][column])

    for i in range(1, len(segments)):
        for j in range(0, i):
            # Перестановочный тест выполняется довольно медленно, можем посмотреть в процессе что происходит
            print(f'Сравниваем "{segments[i]}" c "{segments[j]}" по {column}...')
            stat_res = stats.permutation_test((values_by_seg[i], values_by_seg[j]),
                                 statistic, n_resamples=5000, vectorized = True)
            
            print(f'pvalue = {stat_res.pvalue}, delta = {stat_res.statistic}')
            if stat_res.pvalue < alpha:
                if stat_res.statistic > 0:
                    greater = segments[i]
                    less = segments[j]
                else:
                    greater = segments[j]
                    less = segments[i]
                accepted_hypotheses.append(f'{column_description} в сегменте "{greater}" больше чем в "{less}"')
```

```python
common_accepted = []

compare_segments(common_stats, 'fut_sum', 'Выручка в будущем', common_accepted)
compare_segments(common_stats, 'fut_cnt', 'Число заказов в будущем', common_accepted)

display(common_accepted)
```
