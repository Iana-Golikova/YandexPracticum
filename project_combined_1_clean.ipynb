Добавим столбец с возрастным ограничением.

```python
rating_dict={'AO': 18,
               'E':6,
               'E10+':10,
               'EC':3,
               'M':17,
               'T':13,
               'undefined':float('nan')}
all_rating=prepared_data['rating'].sort_values().unique().tolist()
for i in all_rating:
    prepared_data.loc[prepared_data['rating']==i,'age']=rating_dict[i]
```

Посмотрим, отличается ли рейтинг одной и той же игры на разных платформах.

```python
number_of_games_with_double_rating=0
for name in prepared_data['name'].sort_values().unique():
    rates = prepared_data[prepared_data['name']==name]['rating'].sort_values().unique()
    if len(rates)>1:
        if((rates[0]!='undefined')&(rates[1]!='undefined')):
            number_of_games_with_double_rating+=1
            print(name, rates)
print('Всего игр более с чем одним рейтингом:', number_of_games_with_double_rating)
```

Если бы игр с более чем одним рейтингом было мало, то можно было бы заполнить пропущенные значения рейтином той же игры с других платформ. Однако игр с различными рейтингами на разных платфорах довольно много и восстанавливать пропущенные значения таким образом для всего датасета не очень корректно, при анализе рейтинга выполним исправление только в анализируемой части.

Персональные компьютеры, хоть и сильно изменили свои характеристики с течением времени,  при анализе рынка компьютерных игр не разделяется на версии.

Для более детального представления жизненного цикла платформ создадим сводную таблицу и посчитаем основные метрики.

```python
# cколько лет просуществала каждая платформа
platform_year_set=prepared_data.pivot_table(index=['platform'], values='year_of_release', aggfunc=['nunique','min','max'])
# выведем ее полностью(платформ не так много)
platform_year_set.columns=['nunique','min','max']
platform_year_set['period']=platform_year_set['max']-platform_year_set['min']
#platform_year_set['platform']=platform_year_set.index
platform_year_set=platform_year_set.sort_values('min')
platform_year_set['years_between']=0

# сколько лет была активана
platforms = platform_year_set.index
for i in range(len(platforms)-1):
    platform_year_set.loc[platforms[i+1],'years_between']\
    =platform_year_set.loc[platforms[i+1],'min']\
    -platform_year_set.loc[platforms[i],'min']



platform_sales_per_year_all=prepared_data.pivot_table(index=[ 'year_of_release'],columns=['platform'],
                          values=['total_sales'],
                         aggfunc=['sum'])
platform_sales_per_year_all.columns=list(map(lambda x: x[2], platform_sales_per_year_all.columns))

# год с максимальными продажами
maxes = platform_sales_per_year_all.max()
platform_sales_per_year_t = platform_sales_per_year_all.transpose()
years = platform_sales_per_year_t.columns
platform_year_set['top_year'] = 0
for i in platform_sales_per_year_all.columns:
    platform_year_set.loc[i,'top_year'] =  list(filter(lambda x: platform_sales_per_year_all.loc[x,i]==maxes[i],years))[0]
    
# на какой год существования платформы приходится пик продаж
platform_year_set['number_of_top_year']=platform_year_set['top_year']-platform_year_set['min']

# выведем всю таблицу, платформ не так уж много
print(' Данные о платформах:\n')
display(platform_year_set)


# посчитаем метрики
print('\n Основные метрики\n')
display(platform_year_set.describe())
```

```python
platform_year_set.plot(y='min', ylim=[1978,2016], figsize=(20,20), style='o', grid=True, legend=False,
                       yticks=range(1980,2015), xticks = range(0,31), title= "Год выхода платформы");
plt.xlabel("Платформа")
plt.ylabel("Год");
```

В среднем, новая платформа появляется раз в год,  существует 6-7 лет и достигает своего пика на 2-3 год жизни.

### Анализ влияния оценок пользователей и критиков на продажи

```python
def analyze_dependency(data, platform, score_column, sales_column, score_description):
    subset = data[(data['platform']==platform)&data[score_column].notna()]
    ax = subset.plot(kind = 'scatter', x = score_column, y = sales_column, figsize = (20,15))    
    ax.set_yscale('log')
    plt.xlabel(f'Оценка {score_description}')
    plt.ylabel('Продажи');    
    plt.title(f'Связь продаж с оценкой {score_description}');
    
    print(f'Корреляция оценки {score_description} с продажами для {platform}:',
          subset[score_column].corr(subset[sales_column]))
    print(f'Корреляция оценки {score_description} с логарифмом продаж для {platform}:',
          subset[score_column].corr(np.log(subset[sales_column])))
    print('')
```

```python
analyze_dependency(actual_set,'PC', 'user_score','total_sales', 'пользователей')

analyze_dependency(actual_set,'PC', 'critic_score','total_sales', 'критиков')
```

Оценка пользователей практически никак не связана с продажами, оценка критиков имеет небольшую положительную связь.
При этом корреляция с логарифмом количества продаж заметно выше чем с самим количеством продаж (анализ без логарифмирования по сути учитывает только самые продаваемые игры, а для них связь меньше что видно по диаграмме рассеивания).

```python
# функция, возвращающая сводную таблицу по group_by и index как спискок
def pivot_by_region_sales_top(data, group_by, values_by, n = None):
    pivot_by_region=data.pivot_table(index= group_by,values=values_by, aggfunc=['sum', 'count','mean','median'])
    pivot_by_region.columns=['sales', 'count','mean','median']
    pivot_by_region=pivot_by_region.sort_values('sales', ascending=False)
    keep_top = pivot_by_region.index.to_list()
    if(~(n is None)):
        keep_top = keep_top[:n]
    return (pivot_by_region, keep_top)

# сводная таблица по продажам по платформам и диаграмма, топ 5 платформ
def analyze_platform_sales_by_region(data, group_by, values_by, n, region, flag=True):
    (pivot_by_region, keep_top) = pivot_by_region_sales_top(actual_set, group_by, values_by, n)
    if flag==True:
        print(f'Продажи по платформам в {region}.')
        display(pivot_by_region)
        
    actual_set_with_other_platform= actual_set.copy()
    actual_set_with_other_platform['platform'] = actual_set_with_other_platform['platform']\
    .where(actual_set_with_other_platform['platform'].isin(keep_top),'Other')
    pivot_by_region2=actual_set_with_other_platform.pivot_table(index= group_by,values=values_by,aggfunc=['sum', 'count'])
    pivot_by_region2.columns=['sales', 'count']
    
    diagramm222=pivot_by_region2.plot(y='sales',kind='pie',autopct='%1.2f%%', figsize=(12, 12), legend=True,
                                                shadow=False,startangle=-110,colors=['orange', 'salmon', 'cornflowerblue',
                                                        'palegreen','silver','olive','papayawhip',
                                                        'violet','red','cyan',
                                                        'rosybrown','dimgrey', 'darkgreen']);
    diagramm222.set_ylabel(" ");
    plt.title(f'Продажи по платформам в {region}');
    
    top_five_platforms = pivot_by_region.index.to_list()[:5]
    return top_five_platforms
    
    
   
    
```

```python
def unique_defined(a):
    uni = a.unique()    
    # тут небольшой хак, потому что у обеих игр с двумя разными заполненными 
    # рейтингами более частый идёт вторым
    if (len(uni)>1):
        if (uni[1]!='undefined'):
            return uni[1]
    return uni[0]

### это кривой workaround, но через pandas.DataFrame.rename c мультииндексом совладать не удалось.
def rename_columns(data, columns_dict):
    new_columns = []
    for c in data.columns:
        new_columns.append(columns_dict[c])
    data.columns = new_columns
    
    
def pivot_by_games_in_region(data,sales_region ):
    actual_set_region_games=data.pivot_table(index='name', values=[sales_region,'age', 'rating','genre'],
                                           aggfunc={sales_region:['sum','count'], 'age':['first'], 'rating':[unique_defined],
                                                    'genre':['first']})

    
    rename_columns(actual_set_region_games, 
                {
                 ('age','first'):'age',
                 ('genre','first'):'genre',
                 (sales_region,'count'):'count',
                 (sales_region,'sum'):'sales',
                 ('rating','unique_defined'):'rating'                 
                })
    actual_set_region_games=actual_set_region_games.sort_values(by=('sales'),ascending=False)
    return actual_set_region_games
```

```python
# корреляция между возрастом и продажами
def age_sales_correlation(data,sales_col):
    return data['age'].corr(np.log(data[(data[sales_col]>0)][sales_col]))

```

```python
# хитмапы влияния рейтинга и жанра на продажи 
def rating_heatmap(pivot_by_games,top_list, sales_region, region):
    actual_set_rating_s=pivot_by_games[pivot_by_games['genre'].isin(top_list)]\
                            .pivot_table(index=['rating'], columns='genre', values=sales_region,  aggfunc=['sum'])
    actual_set_rating_m=pivot_by_games[pivot_by_games['genre'].isin(top_list)]\
                            .pivot_table(index=['rating'], columns='genre', values=sales_region,  aggfunc=['mean'])
    print(f'Влияние рейтинга и жанра на суммарные/средние продажи в {region}\n')
    fig, (ax1, ax2) = plt.subplots(ncols=2)
    fig.subplots_adjust(wspace=0.01)   
    sns.heatmap(actual_set_rating_s, annot=True, annot_kws={'size':12}, cmap="Reds", ax=ax1);    
    sns.heatmap( actual_set_rating_m, annot=True, annot_kws={'size':12}, cmap="Reds", ax=ax2);
```

## Проверка гипотез

### Гипотеза о равенстве средних пользовательских рейтинги платформ Xbox One и PC 

Для проверки этой гипотезы отберем только те игры, которые представленны на обеих платформах.  Так как такие выборки зависмимы, используем двустроннюю гипотезу для провекри равенства среднего двух зависимых совокупностей. 

```python

pivot_game_platform = actual_set.pivot_table(index = 'name', columns = 'platform', values = 'user_score', aggfunc = 'first')
pivot_game_platform = pivot_game_platform[['PC', 'XOne']]
pivot_game_platform = pivot_game_platform[(pivot_game_platform['PC']>0)&(pivot_game_platform['XOne']>0)]
display(pivot_game_platform)

XOne_user_score=pivot_game_platform['PC']
PC_user_score=pivot_game_platform['XOne']
#XOne_user_score=actual_set[(actual_set['platform']=='XOne')&actual_set['user_score'].notna()]['user_score']
print('Матожидание оценки игр пользователями платформы XOne ', XOne_user_score.mean())
#PC_user_score=actual_set[(actual_set['platform']=='PC')&actual_set['user_score'].notna()]['user_score']
print('Матожидание оценки игр пользователями платформы PC', PC_user_score.mean())
```

Нулевая гипотеза: матожидания оценок одних и тех же игр на PC и XOne равны.

Альтернатива: матожидания не равны. 

```python
alpha = 0.01 # уровень статистической значимости
# если p-value окажется меньше него, отвергнем гипотезу

results = st.ttest_rel(XOne_user_score,PC_user_score)

print('p-значение:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу') 
```

Вывод: Пользователи PC в среднем выше оценивают игры, чем пользовати XOne.

```python
print('Корреляция между оценками на двух платформах для игр присутствующих на обеих:',
      XOne_user_score.corr(PC_user_score))
```

```python
XOne_user_score_full=actual_set[(actual_set['platform']=='XOne')&actual_set['user_score'].notna()]['user_score']
PC_user_score_full=actual_set[(actual_set['platform']=='PC')&actual_set['user_score'].notna()]['user_score']
print('Матожидание и дисперсия оценки игр пользователями платформы XOne ',
      XOne_user_score_full.mean(), XOne_user_score_full.var())

print('Матожидание и дисперсия оценки игр пользователями платформы PC', 
      PC_user_score_full.mean(), PC_user_score_full.var())
```

Нулевая гипотеза: матожидания оценок пользователей на PC и XOne равны.

Альтернатива: матожидания не равны. 

```python
alpha = 0.01 # уровень статистической значимости
# если p-value окажется меньше него, отвергнем гипотезу

results = st.ttest_ind(XOne_user_score_full, PC_user_score_full)#, equal_var=False)

print('p-значение:', results.pvalue)

if results.pvalue < alpha:
    print('Отвергаем нулевую гипотезу')
else:
    print('Не получилось отвергнуть нулевую гипотезу') 
```

В среднем, пользователи PC и XOne одинаково оценивают игры.
